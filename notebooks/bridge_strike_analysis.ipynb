# NYC Bridge-Strike Risk Analysis

Interactive walkthrough of the pipeline in `src/bridge_strikes/`. All logic lives in the package; this notebook only runs it and displays results.

Setup: from the repo root, run `pip install -r requirements.txt`, then pick the `.venv` interpreter as the notebook kernel.

```python
import logging
%matplotlib inline
logging.basicConfig(level=logging.INFO, format='[%(levelname)s] %(name)s: %(message)s')

from bridge_strikes.pipeline import run_pipeline, print_summary
from bridge_strikes.viz import plot_model_comparison, plot_monthly_series, plot_top_risk_locations
```

## 1. Run the pipeline
Uses cached CSVs in `data/raw/` if present; otherwise downloads from NYC Open Data. Pass `refresh=True` to force a re-download.

```python
result = run_pipeline(refresh=False, save=True)
print_summary(result)
```

## 2. Monthly strike counts

```python
plot_monthly_series(result.monthly);
```

## 3. Poisson vs Negative Binomial
Both models are fit by maximum likelihood in statsmodels, so their AIC values are comparable. A variance/mean ratio well above 1 in the monthly counts favours the Negative Binomial.

```python
display(result.comparison)
plot_model_comparison(result.comparison);
```

## 4. Location-level risk
`p_next_month` is a Gamma-Poisson estimate using the full study window as exposure.

```python
display(result.location_risk.head(10))
plot_top_risk_locations(result.location_risk, top_n=10);
```
